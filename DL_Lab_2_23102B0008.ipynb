```python

from itertools import product

print("Select Gate:")
print("1. AND")
print("2. OR")
print("3. XOR")
print("4. NOT")

choice = int(input("Enter your choice (1/2/3/4): "))

# NOT gate has only one input
if choice == 4:
    n = 1
else:
    n = int(input("Enter number of inputs: "))

# Display activation function
print("\nActivation Function Used")

if choice == 1:
    print(f"Threshold Function: Yout = 1 if Yin == {n} else 0")
elif choice == 2:
    print("Threshold Function: Yout = 1 if Yin >= 1 else 0")
elif choice == 3:
    print("Modulus Function: Yout = 1 if Yin % 2 == 1 else 0 (Odd parity)")
elif choice == 4:
    print("Complement Function: Yout = NOT x1")
else:
    print("Invalid Choice")
    exit()

# Header
for i in range(n):
    print(f"x{i+1}", end="\t")
print("Yin\tYout")

# Generate all input combinations
for inputs in product([0, 1], repeat=n):

    yin = sum(inputs)

    if choice == 1:      # AND
        yout = 1 if yin == n else 0

    elif choice == 2:    # OR
        yout = 1 if yin >= 1 else 0

    elif choice == 3:    # XOR
        yout = 1 if yin % 2 == 1 else 0

    elif choice == 4:    # NOT
        yin = inputs[0]
        yout = 1 if yin == 0 else 0

    for bit in inputs:
        print(bit, end="\t")
    print(yin, "\t", yout)
```
